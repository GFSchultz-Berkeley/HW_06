````python
# Q6-2 Grading Tag:
scores = { "a": 1, "c": 3, "b": 3, "e": 1,  "d": 2, "g": 2,
           "f": 4, "i": 1, "h": 4, "k": 5,  "j": 8, "m": 3,
           "l": 1, "o": 1, "n": 1, "q": 10, "p": 3, "s": 1,
           "r": 1, "u": 1, "t": 1, "w": 4,  "v": 4, "y": 4,
           "x": 8, "z": 10 }
 
WILDCARD = "*"
MIN_RACK_SIZE = 2
MAX_RACK_SIZE = 7
MAX_WILDCARDS = 2
ALLOWED_LETTERS = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

def score_word(word: str) -> int:
    # TODO: Implement score_word here.
    #raise NotImplementedError(
        #"Implement this function to score a word based on the scores dictionary."
    #)
    total = 0
    for char in word:
        total += scores.get(char.lower(), 0)
    return total


def validate_rack(rack: str) -> bool:
    # TODO: Implement validate_rack here.
    #raise NotImplementedError(
        #"Implement this function to validate the rack against the rules in the description."
    #)
    if not isinstance(rack, str):
        return False

    if not MIN_RACK_SIZE <= len(rack) <= MAX_RACK_SIZE:
        return False

    for char in rack.upper():
        if char != WILDCARD and char not in ALLOWED_LETTERS:
            return False

    return rack.count(WILDCARD) <= MAX_WILDCARDS


def run_scrabble(rack: str, words: list[str]) -> dict[str, int]:
    # TODO: Implement run_scrabble here.
    #raise NotImplementedError(
        #"Implement this function to find all valid words and their scores based on the rack provided."
    #)
    rack = rack.upper()
    rack_size = len(rack)
    wildcards = rack.count(WILDCARD)

    rack_counts = {}
    for title in rack:
        if title != WILDCARD:
            rack_counts[title] = rack_counts.get(title, 0) + 1

    results = {}

    for raw_word in words:
        word = raw_word.strip().upper()

        if not word or len(word) > rack_size:
            continue
        available = rack_counts.copy()
        wildcards_left = wildcards

        played = []
        playable = True

        for letter in word:
            if available.get(letter, 0) > 0:
                available[letter] -= 1
                played.append(letter)
            elif wildcards_left > 0:
                wildcards_left -= 1
                played.append(WILDCARD)
            else:
                playable = False
                break

        if playable:
            results[word] = score_word("".join(played))

        return results


if __name__ == "__main__":
    # The condition above checks if this script is being run directly or being imported. More info:
    # https://docs.python.org/3/library/__main__.html
    # https://realpython.com/if-name-main-python/#in-short-it-allows-you-to-execute-code-when-the-file-runs-as-a-script-but-not-when-its-imported-as-a-module
    words = read_sowpods(SOWPODS_PATH)
    print(f"Read {len(words)} words from SOWPODS file.")
    rack = "EXAMPLE"  # feel free to use input() to get the rack from the user.

    # TODO: Call the functions above to complete the scrabble solver

    if not validate_rack(rack):
        print(f"'{rack}' is not a valid rack. Use 2-7 letters, with at most two '*' wildcards.")
    else:
        results = run_scrabble(rack, words)
        if not results:
            print(f"No words can be made from '{rack}'.")
        else:
            ranked = sorted(results.items(), key=lambda item: (-item[1], item[0]))
            for word, score in ranked:
                print(f"{score:>3} {word}")
            print(f"\n{len(results)} words found for rack '{rack}'.")
````
